# Visualization

## Import libraries

```python
import os
import pandas as pd

from visualizer.visualizer import Visualizer
from utils.constants import SCRAPER_START_DATE, SCRAPER_END_DATE
from utils.enums import GenerateDateTimeType
```

```python
my_visualizer = Visualizer()
my_visualizer.connect_to_database(database_name=os.getenv("SILVER_POSTGRES_DATABASE"))
```

## silver_gdp

```python
silver_gdp = my_visualizer.select(schema_name="macroeconomics", table_name="gdp")
silver_gdp
```

```python
std_silver_gdp = my_visualizer.make_date_time_index_for_dataframe(silver_gdp)
std_silver_gdp
```

```python
std_silver_gdp_fig = my_visualizer.plot_line_chart(
    std_silver_gdp,
    x_column="date",
    y_columns=["gdp_growth"],
    title="GDP Growth",
    y_axis_title="Rate",
    y_unit="%",
    prefix_figure_name="gdp",
)
```
