# Create PDF

## Import libraries

```python
from PIL import Image
import os
import random

from utils.constants import (
    CHARTS_DIR_BASE,
    CHARTS_DIR_MACROECONOMICS,
    CHARTS_DIR_STOCK_MARKET,
    CHARTS_DIR_ENTERPRISE,
    PDF_OUTPUT_DIR
)
```

## Parameters

```python
print(CHARTS_DIR_BASE)
print(CHARTS_DIR_MACROECONOMICS)
print(CHARTS_DIR_STOCK_MARKET)
print(CHARTS_DIR_ENTERPRISE)
print(PDF_OUTPUT_DIR)
```

## Export MACROECONOMICS charts to PDF

```python
os.makedirs(PDF_OUTPUT_DIR, exist_ok=True)

base_path = CHARTS_DIR_MACROECONOMICS
file_name = "macroeconomics"
pdf_output_dir = PDF_OUTPUT_DIR

# Get all PNG files in the folder
all_png_files = [f for f in os.listdir(base_path) if f.lower().endswith(".png")]

# Build full paths
png_paths = [os.path.join(base_path, fname) for fname in all_png_files]

# Open and convert to RGB
image_list = [Image.open(png).convert("RGB") for png in png_paths]

# Save as PDF
output_path = os.path.join(pdf_output_dir, f"{file_name}.pdf")
image_list[0].save(output_path, save_all=True, append_images=image_list[1:])

print("Selected files:", all_png_files)
print(f"PDF saved at: {pdf_output_dir}")
```
