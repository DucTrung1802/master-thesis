## ENTERPRISE

```python
my_visualizer.connect_to_database(database_name=os.getenv("BRONZE_POSTGRES_DATABASE"))
my_visualizer.set_chart_dir(CHARTS_DIR_ENTERPRISE)
```

```python
def get_code_table_and_plot(code: str):
    code = code.upper()

    if not code:
        print("Need to specify enterprise code before plotting.")
    else:
        gold_enterprise = my_visualizer.select(
            schema_name="enterprise",
            table_name="daily_price",
            conditions=[
                Condition(
                    column=Table.DAILY_PRICE.Column.CODE.value,
                    operator=SqlOperator.EQUAL_TO,
                    value=code,
                    data_type=DataType.VARCHAR,
                )
            ],
        )

        if len(gold_enterprise) == 0:
            print(f"No data found for code: {code}.")
            return

        gold_enterprise = gold_enterprise.astype(
            {
                Table.DAILY_PRICE.Column.DATE.value: "datetime64[ns]",
                Table.DAILY_PRICE.Column.CODE.value: "string",
                Table.DAILY_PRICE.Column.MARKET_ID.value: "int",
                **{
                    col: "float"
                    for col in gold_enterprise.columns
                    if col
                    not in [
                        Table.DAILY_PRICE.Column.DATE.value,
                        Table.DAILY_PRICE.Column.CODE.value,
                        Table.DAILY_PRICE.Column.MARKET_ID.value,
                    ]
                },
            }
        )

        display(gold_enterprise)

        gold_enterprise_fig = my_visualizer.plot_lines_bars_chart(
            gold_enterprise,
            x_column="date",
            y_line_columns=["close"],
            y_bar_columns=["volume"],
            title=f"{code} close",
            y_line_axis_title="Point",
            y_bar_axis_title="",
            y_unit="",
            prefix_figure_name=f"{code.lower()}",
            marker=None,
        )
        
        display(gold_enterprise_fig)
```
